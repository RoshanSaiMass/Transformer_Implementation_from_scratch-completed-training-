```python
#In multi-head attention there is not much of things to do except the fact that we have to have multiple SA blocks.
```

```python
class SelfAttention(nn.Module):
  def __init__(self,embed_size,heads):
    super().__init__()
    self.embed_size=embed_size
    self.Q=nn.Linear(embed_size,embed_size)
    self.K=nn.Linear(embed_size,embed_size)
    self.V=nn.Linear(embed_size,embed_size)
  def forwardpass(self,embedding):
    self.q=self.Q(embedding)
    self.k=self.K(embedding)
    self.v=self.V(embedding)
    print(f"The radomly allocated Q,K and V values are the following : {self.q},{self.k},{self.v}")
  def coeff(self):
    self.coeff=torch.softmax(self.q@self.k.T,dim=1)
    print(f"The coeff or the final weights we get to multiply are:{self.coeff}")
  def finalemb(self):
    self.finalemb=self.coeff@self.v
    print(f"The final embedding is:{self.finalemb}")
```

```python
class MultiHeadAttention(nn.Module):
  def __init__(self,embed_size,heads):
    super().__init__()
    self.embed_size=embed_size
    self.heads=heads
    self.sa=[]
    for i in range(heads):
      self.sa.append(SelfAttention(embed_size,heads))
  def forward(self,embedding):
    for i in range(self.heads):
      self.sa[i].forwardpass(embedding)
  def coeffs(self):
    for i in range(self.heads):
      self.sa[i].coeff()
  def finalembs(self):
    for i in range(self.heads):
      self.sa[i].finalemb()
```
